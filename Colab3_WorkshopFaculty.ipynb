## ==============================================================================
## MODULE 2: RECURRENT EMBEDDED INTENT DECODER (LAYER 2: COMPUTE CORE)
## ==============================================================================

```python
# === MODULE 2: LAYER 2: COMPUTE CORE ===
class MicroBMIController:
    def __init__(self, buffer_size=5):
        self.buffer_size = buffer_size
        self.buffer = [0] * buffer_size
        
    def add_sample(self, sample):
        # Shift contents to maintain a fixed-width rolling window
        self.buffer.pop(0)
        self.buffer.append(sample)
        
    def compute_embedded_intent(self, manual_offset_shift=3):
        """Processes the rolling buffer natively using raw integer arithmetic."""
        buffer_sum = sum(self.buffer)
        smoothed_average = buffer_sum // self.buffer_size
        
        # Micro-scale decoding transformation to map neural signal to a 5-step spatial coordinate
        decoded_coordinate = (smoothed_average - manual_offset_shift) // 5
        
        # Constrain output to a hard physical boundary (0 to 4 grid coordinates)
        return max(0, min(4, decoded_coordinate))
```
