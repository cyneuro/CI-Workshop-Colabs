#### 📌 Reference Implementation – SSH Configuration Generator
*Safe to run in Colab; it only prints text. Create the real file inside FABRIC's JupyterHub.*

```python
#@title ▶ Generate an ~/.ssh/config template  (reference – for study)
GATEWAY_USERNAME = "your_fabric_gateway_username"  #@param {type:"string"}
GATEWAY_KEY       = "/home/fabric/work/fabric_config/fabric_bastion_key"  #@param {type:"string"}
SESSION_KEY       = "/home/fabric/work/fabric_config/slice_key"           #@param {type:"string"}

print(f'''# ~/.ssh/config  (place inside FABRIC\'s JupyterHub)
# "bastion.fabric-testbed.net" is the secure gateway computer every connection passes through first.
Host bastion.fabric-testbed.net
    User {GATEWAY_USERNAME}
    IdentityFile {GATEWAY_KEY}
    ForwardAgent yes

# ProxyJump automatically routes every other connection through the secure gateway above.
Host * !bastion.fabric-testbed.net
    ProxyJump {GATEWAY_USERNAME}@bastion.fabric-testbed.net:22
    IdentityFile {SESSION_KEY}
    User ubuntu
    StrictHostKeyChecking no
    UserKnownHostsFile /dev/null
''')
print("Then run inside JupyterHub:  chmod 600 ~/.ssh/config")
print("(chmod 600 restricts the file so only you can read it — required for SSH to trust the key.)")
```

#### 📌 Official FABRIC Computing Session States (Reference Table)

"Ready" means the request finished successfully and every part of the session is working. "Not Ready" means the request finished, but something failed and needs to be inspected.

| Plain-language state | FABRIC's technical name | Meaning |
|---|---|---|
| **Requested** | Nascent | New session, just created. |
| **Setting Up** | Configuring | Resources are being assigned and started. |
| **Ready** | StableOK | Everything requested is working. Ready for connection and scientific work. |
| **Not Ready** | StableError | Some resources started, others failed. Inspect which ones failed. |
| **Update Successful** | ModifyOK | A change to the session succeeded. |
| **Update Failed** | ModifyError | A change partially succeeded; some resources failed. |

> You must wait until a computing session reaches **Ready** before attempting to connect or run experiments.

## 📖 Reference Guide: Key Concepts for Accessing National Research Computing Resources

### 🌐 FABRIC & Testbed Fundamentals (Concepts 1–10)
1. **Research Testbed** – A shared, programmable facility (FABRIC, CloudLab, …) where researchers can run real computing and networking experiments at national scale.
2. **Computing Session** – A temporary, isolated bundle of resources allocated to one experiment (FABRIC calls this a "slice").
3. **Secure Gateway Computer** – A hardened server (also called a "bastion host") that every connection must authenticate through first, before reaching any research computing node.
4. **Resource Instance** – One individual piece of a computing session — for example, one virtual computer or one network link.
5. **Session State** – The life-cycle stage of a computing session: Requested → Setting Up → Ready (or Not Ready). Only "Ready" sessions are available for research work.
6. **fablib** – FABRIC's official Python toolkit for requesting, managing, and releasing computing sessions.
7. **Management Address** – The network address, reachable through the secure gateway, used to connect to and control a computing node.
8. **ProxyJump** – An SSH setting that automatically routes a connection through the secure gateway computer, so you do not have to do it manually every time.
9. **Resource Request** – The processor cores, working memory (RAM), permanent storage, and starting image a researcher asks for when defining a node.
10. **Session Cleanup** – The mandatory final step of releasing resources so they return to the shared pool for other researchers.

### 💻 Computing & Workflow Fundamentals (Concepts 11–15)
11. **Temporary Cloud Workspace** – A computing session is not permanent storage; anything saved only on the session disappears once the session is released. Save your results elsewhere first.
12. **Secure Key Hygiene** – Never share a private key with anyone. Restrict key files with `chmod 600`, and keep your secure-gateway key and computing-session key separate.
13. **Reproducible Experiment Description** – Writing your resource request and setup as code (rather than clicking through a series of menus) means anyone, including future you, can recreate the exact same computing session.
14. **Community Resource Ethics** – National computing systems are shared. Request only what your research actually needs, and release it promptly when finished.
15. **From Simulation to Production** – The request → submit → inspect → release pattern you practiced in Colab maps directly onto the real `fablib` commands you will run inside FABRIC's JupyterHub.
