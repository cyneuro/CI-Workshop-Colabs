### 🧠 MODULE 0A: Frequency Filtering & Artifact Mitigation
#### 🦾 CELL A.1: The Digital Butterworth Bandpass & Artifact Filter Solution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt

# 1. Setup Signal parameters
fs = 500  # 500 Hz sampling rate
t = np.linspace(0, 5, fs * 5, endpoint=False)

# Nyquist comment: Nyquist Frequency = fs / 2 = 250 Hz. All tracked components must stay below this ceiling.
alpha = np.sin(2 * np.pi * 10 * t)  # Target 10 Hz alpha wave rhythm
noise_60hz = 0.8 * np.sin(2 * np.pi * 60 * t)  # Electrical line noise
raw_signal = alpha + noise_60hz

# Inject physical blink/muscle jerk spike artifact
raw_signal[600:630] += 8.0

# 2. Design Butterworth Bandpass Filter (1-30 Hz)
lowcut, highcut = 1.0, 30.0
nyq = 0.5 * fs
low = lowcut / nyq
high = highcut / nyq
b, a = butter(4, [low, high], btype='band')
filtered_signal = filtfilt(b, a, raw_signal)

# 3. Threshold Detector for Artifact Flagging
threshold = 3.5
artifacts = np.abs(filtered_signal) > threshold
cleaned_signal = filtered_signal.copy()
cleaned_signal[artifacts] = 0.0  # Zero-out/clip artifact regions cleanly

# Plot
fig, axes = plt.subplots(3, 1, figsize=(10, 6))
axes[0].plot(t, raw_signal, color='red', label='Raw Signal (Alpha + 60Hz + Spike)')
axes[0].set_title("Raw Corrupted Stream")
axes[0].legend()

axes[1].plot(t, filtered_signal, color='orange', label='Butterworth Filtered (1-30 Hz)')
axes[1].set_title("Frequency Filtering Isolation")
axes[1].legend()

axes[2].plot(t, cleaned_signal, color='green', label='Cleaned Vector Array')
axes[2].set_title("Final Artifact Rejected Array")
axes[2].legend()
plt.tight_layout()
plt.show()
```

#### 🧠 CELL A.2: Time-Frequency Spectrogram Generation Solution

```python
from scipy.signal import stft

# 1. Create a signal where the dominant frequency shifts over time
fs_stft = 200
t_stft = np.linspace(0, 5, fs_stft * 5, endpoint=False)
signal_stft = np.zeros_like(t_stft)
signal_stft[:2*fs_stft] = np.sin(2 * np.pi * 8 * t_stft[:2*fs_stft])  # 8 Hz alpha for 2s
signal_stft[2*fs_stft:] = np.sin(2 * np.pi * 25 * t_stft[2*fs_stft:]) # 25 Hz beta for 3s

# 2. Compute Short-Time Fourier Transform (STFT)
frequencies, times, z_stft = stft(signal_stft, fs=fs_stft, nperseg=64)

# 3. Plot Spectrogram
plt.figure(figsize=(8, 4))
plt.pcolormesh(times, frequencies, np.abs(z_stft), shading='gouraud', cmap='magma')
plt.title('Time-Frequency Power Spectral Density Spectrogram')
plt.ylabel('Frequency (Hz)')
plt.xlabel('Time (seconds)')
plt.colorbar(label='Power Intensity')
plt.ylim(0, 40)
plt.show()
```
