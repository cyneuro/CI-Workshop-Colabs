## ==============================================================================
## STAGE 1: OBTAINING DATA
## ==============================================================================

```python
# === STAGE 1: OBTAINING DATA ===
import time
import json
import sys
import numpy as np
import matplotlib.pyplot as plt

def stage_1_obtaining_data():
    print("\n[STAGE 1: OBTAINING DATA] Accessing open research dataset...")
    np.random.seed(42)
    
    # Generate 100 neural tracking channels across 500 temporal bins
    mock_spikes = np.random.poisson(lam=0.3, size=(100, 500))
    
    # Pack experimental data and descriptive details
    neural_payload = {
        "matrix": mock_spikes,
        "metadata": {
            "subject_id": "SUBJ-01-LOCAL",
            "electrode_array_type": "High-Density Extracellular Grid",
            "temporal_resolution_ms": 20
        }
    }
    print(f"[STAGE 1] Data access complete. Data dimensions: {mock_spikes.shape}")
    return neural_payload
```

## ==============================================================================
## STAGE 2: ANALYZING DATA
## ==============================================================================

```python
# === STAGE 2: ANALYZING DATA ===
def stage_2_analyzing_data(neural_payload, analysis_sensitivity=0.52):
    print("[STAGE 2: ANALYZING DATA] Applying processing and movement analysis methods...")
    matrix = neural_payload["matrix"]
    
    # Static tracking coefficients
    w_x = np.random.normal(0, 0.1, size=(matrix.shape[0],))
    w_y = np.random.normal(0, 0.1, size=(matrix.shape[0],))
    
    pred_x = np.dot(w_x, matrix) * analysis_sensitivity
    pred_y = np.dot(w_y, matrix) * analysis_sensitivity
    
    print(f"[STAGE 2] Analysis complete. Estimated movement pathways generated.")
    return pred_x, pred_y
```

## ==============================================================================
## STAGE 3: VISUALIZING RESULTS
## ==============================================================================

```python
# === STAGE 3: VISUALIZING RESULTS ===
def stage_3_visualizing_results(pred_x, pred_y):
    print("[STAGE 3: VISUALIZING RESULTS] Creating figures and interactive displays...")
    
    # Calculate accuracy error relative to an ideal central target
    system_mse = float(np.mean(pred_x**2 + pred_y**2))
    
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.plot(pred_x, pred_y, label='Decoded Movement Pathway', color='navy', alpha=0.8)
    ax.scatter(0, 0, color='red', marker='X', s=150, label='Target Focus')
    
    ax.set_title("Neural Signal Decoded Spatial Map")
    ax.set_xlabel("Horizontal Position Axis ($\Delta$X)")
    ax.set_ylabel("Vertical Position Axis ($\Delta$Y)")
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='upper right')
    
    plt.tight_layout()
    plt.close()
    print(f"[STAGE 3] Visual display complete. Pathway Variance (MSE): {system_mse:.4f}")
    return fig, system_mse
```

## ==============================================================================
## STAGE 4: DOCUMENTING RESEARCH
## ==============================================================================

```python
# === STAGE 4: DOCUMENTING RESEARCH ===
def stage_4_documenting_research(subject_metadata, sensitivity_used, error_metric):
    print("[STAGE 4: DOCUMENTING RESEARCH] Capturing analysis settings and documentation details...")
    
    reproducibility_manifest = {
        "timestamp_epoch": time.time(),
        "workflow_notebook_target": "Colab 1: Building Your First Research Workflow",
        "configured_parameters": {
            "applied_analysis_sensitivity": sensitivity_used,
            "accuracy_threshold_limit": 0.85
        },
        "extracted_features": {
            "calculated_pathway_mse": error_metric
        },
        "subject_assignment": subject_metadata,
        "workflow_status_verdict": "VERIFIED_PASS" if error_metric <= 0.85 else "ACCURACY_DRIFT_WARNING"
    }
    
    output_filename = "colab1_reproducibility_manifest.json"
    with open(output_filename, 'w') as json_file:
        json.dump(reproducibility_manifest, json_file, indent=4)
        
    print(f"[STAGE 4] Research documentation log saved to: '{output_filename}'")
    return reproducibility_manifest
```

## ==============================================================================
## STAGE 5: LABORATORY PLAYGROUND
## ==============================================================================
### FACULTY KEY: Orchestration Engine and Testing Scenarios

```python
# === STAGE 5: LABORATORY PLAYGROUND ===
def run_laboratory_playground_experiment(upstream_analysis_sensitivity=0.52):
    """
    Central orchestration runner representing the instructor master solution playground for Colab 1.
    """
    print("\n" + "="*70)
    print(f"RUNNING LABORATORY PLAYGROUND EXPERIMENT (Sensitivity Setting = {upstream_analysis_sensitivity})")
    print("="*70)
    
    # Stage 1: Obtaining Data
    neural_payload = stage_1_obtaining_data()
    
    # Stage 2: Analyzing Data
    pred_x, pred_y = stage_2_analyzing_data(neural_payload, analysis_sensitivity=upstream_analysis_sensitivity)
    
    # Stage 3: Visualizing Results
    fig, system_mse = stage_3_visualizing_results(pred_x, pred_y)
    
    # Stage 4: Documenting Research
    manifest = stage_4_documenting_research(neural_payload["metadata"], upstream_analysis_sensitivity, system_mse)
    
    print("\n[PLAYGROUND RECORDED METADATA]:")
    print(json.dumps(manifest, indent=2))
    return manifest

# =====================================================================
# EVALUATION RUNS
# =====================================================================
# Scenario A: Running within normal expected settings (Passes workflow check)
run_laboratory_playground_experiment(upstream_analysis_sensitivity=0.52)

# Scenario B: Modifying settings past limits to trigger an Accuracy Drift warning
run_laboratory_playground_experiment(upstream_analysis_sensitivity=0.95)
```
