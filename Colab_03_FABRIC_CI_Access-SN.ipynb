### 🧭 Step 1.5 — Install FABRIC's Toolkit and Test Your Connection 🪐

This is the first cell that actually talks to FABRIC. It installs `fablib` (FABRIC's official Python toolkit), points it at your token, and asks FABRIC to confirm it recognizes you. This can take a minute or two the first time.

```python
# 🪐 Step 1.5 — Install and configure fablib, then test your connection
!pip install --quiet fabrictestbed-extensions

import os

os.environ["FABRIC_TOKEN_LOCATION"] = TOKEN_PATH
os.environ["FABRIC_PROJECT_ID"] = fabric_project_name

try:
    from fabrictestbed_extensions.fablib.fablib import FablibManager as fablib_manager
    fablib = fablib_manager()
    fablib.show_config()
    print("\n✅ Connected! fablib recognizes your credentials.")
except FileNotFoundError:
    print("⚠️ fablib couldn't find your token file. Re-check Step 1.4 — the file needs to be")
    print(f"   at exactly: {TOKEN_PATH}")
except Exception as e:
    msg = str(e)
    print(f"⚠️ Connection failed: {msg}")
    if "project" in msg.lower():
        print("   This usually means your Project Lead hasn't approved your project membership yet.")
        print("   Contact your workshop organizer and try again once you're approved.")
    else:
        print("   Double-check your token is valid and not expired, then re-run this cell.")
        print("   If the problem continues, see the External Resources section at the end of this notebook.")
```

## 💻 Step 2: Provision & Run Your Computing Session


### 🧭 Step 2.1 — Where Is Your Code?

Choose how you'd like to bring your own code onto the computing session you're about to request. Both options are fully supported below.

```python
# 🧭 Step 2.1 — Choose your code source
code_source = "Google Drive folder"  #@param ["Google Drive folder", "GitHub repository URL"]

# Fill in ONLY the field that matches your choice above.
drive_folder_path = "/content/drive/MyDrive/FABRIC_Workshop/my_code"   #@param {type:"string"}
github_repo_url    = "https://github.com/your-username/your-repo.git"  #@param {type:"string"}
entry_script        = "run.py"  #@param {type:"string"}

print(f"✅ Code source set to: {code_source}")
if code_source == "Google Drive folder":
    print(f"   Will upload the contents of: {drive_folder_path}")
else:
    print(f"   Will clone: {github_repo_url}")
print(f"   Entry script to run on the computing session: {entry_script}")
```

---

### 🧭 Step 2.2 — Request Your Computing Session 🪐

This asks FABRIC for one computing node with modest, workshop-appropriate defaults (2 processor cores, 8 GB working memory, 10 GB storage, Ubuntu 22.04). You can change these if you know you need more.

```python
# 🪐 Step 2.2 — Define and submit a one-node computing session
SLICE_NAME = f"Workshop-{fabric_username.split('@')[0]}"  #@param {type:"string"}
CORES = 2      #@param {type:"integer"}
RAM_GB = 8     #@param {type:"integer"}
DISK_GB = 10   #@param {type:"integer"}

try:
    slice = fablib.new_slice(name=SLICE_NAME)
    site = fablib.get_random_site()
    slice.add_node(name="Node1", site=site, cores=CORES, ram=RAM_GB, disk=DISK_GB, image="default_ubuntu_22")
    slice.submit()
    print(f"✅ Requested computing session '{SLICE_NAME}' at site {site}. Continue to Step 2.3 to wait for it to become Ready.")
except Exception as e:
    print(f"⚠️ Request failed: {e}")
    print("   Common causes: not yet approved on the project, or the chosen site is temporarily full.")
    print("   Re-run this cell to try a different (randomly chosen) site, or contact your organizer.")
```

### 🧭 Step 2.3 — Wait Until Your Session Is Ready 🪐

Computing sessions take a minute or two to set up. This cell checks the state every 15 seconds until it reaches **Ready**, and explains what to do if it doesn't.

```python
# 🪐 Step 2.3 — Poll until the session is Ready (or report a problem)
import time

TIMEOUT_SECONDS = 600
waited = 0

try:
    slice = fablib.get_slice(name=SLICE_NAME)
    while waited < TIMEOUT_SECONDS:
        state = slice.get_state()
        print(f"⏳ [{waited}s] Current state: {state}")
        if state == "StableOK":
            print("✅ Ready! Continue to Step 2.4.")
            break
        if state == "StableError":
            print("⚠️ Not Ready — some part of the request failed. See the state table in the Reference")
            print("   Guide below, then try Step 2.2 again, or contact your organizer if it keeps failing.")
            break
        time.sleep(15)
        waited += 15
    else:
        print("⚠️ Timed out waiting. Re-run this cell to keep checking — FABRIC is likely just busy.")
except Exception as e:
    print(f"⚠️ Could not check status: {e}")
```

### 🧭 Step 2.4 — Send Your Code to the Computing Session and Run It 🪐

This uses the code source you chose in Step 2.1. It uploads (or clones) your code onto the node, then runs your entry script and shows you the output.

```python
# 🪐 Step 2.4 — Upload/clone code and run it on the node
try:
    slice = fablib.get_slice(name=SLICE_NAME)
    node = slice.get_node(name="Node1")

    if code_source == "Google Drive folder":
        node.upload_directory(drive_folder_path, "/home/ubuntu/my_code")
        print(f"✅ Uploaded {drive_folder_path} to the computing session.")
    else:
        stdout, stderr = node.execute(f"git clone {github_repo_url} /home/ubuntu/my_code")
        print(f"✅ Cloned {github_repo_url} on the computing session.")

    print(f"\n▶️ Running {entry_script} ...\n")
    stdout, stderr = node.execute(f"cd /home/ubuntu/my_code && python3 {entry_script}")
    print("----- Output -----")
    print(stdout)
    if stderr:
        print("----- Errors (if any) -----")
        print(stderr)
except Exception as e:
    print(f"⚠️ Something went wrong sending or running your code: {e}")
    print("   Double-check the path/URL you entered in Step 2.1, and that your entry script filename is correct.")
```

### 🧭 Step 2.5 — Retrieve Your Results 🪐

If your script wrote output files, this brings them back to your Colab environment (and, if Drive is connected, saves a copy there too).

```python
# 🪐 Step 2.5 — Download results from the computing session
RESULTS_REMOTE_PATH = "/home/ubuntu/my_code/results"   #@param {type:"string"}
RESULTS_LOCAL_PATH   = "/content/fabric_results"         #@param {type:"string"}

import os
os.makedirs(RESULTS_LOCAL_PATH, exist_ok=True)

try:
    node = fablib.get_slice(name=SLICE_NAME).get_node(name="Node1")
    node.download_directory(RESULTS_LOCAL_PATH, RESULTS_REMOTE_PATH)
    print(f"✅ Downloaded results to {RESULTS_LOCAL_PATH}")
    if drive_available:
        import shutil
        drive_copy = f"{DRIVE_PROFILE_DIR}/results"
        shutil.copytree(RESULTS_LOCAL_PATH, drive_copy, dirs_exist_ok=True)
        print(f"✅ Also saved a copy to Google Drive: {drive_copy}")
except Exception as e:
    print(f"ℹ️ Nothing downloaded ({e}). If your script didn't write files to {RESULTS_REMOTE_PATH}, this is expected.")
```

### 🧭 Step 2.6 — Release Your Computing Session 🪐

> ⚠️ **This step is required, not optional.** FABRIC is a shared national resource — leaving a session running after you're done takes resources away from other researchers, including other workshop participants. Run the cell below now.

```python
# 🪐 Step 2.6 — Release resources back to the shared pool (REQUIRED)
try:
    fablib.get_slice(name=SLICE_NAME).delete()
    print(f"✅ Released '{SLICE_NAME}'. Thank you for cleaning up after yourself!")
except Exception as e:
    print(f"⚠️ Could not release automatically: {e}")
    print("   Log into the FABRIC Portal and delete the slice manually to be safe.")
```

```python
# 🪐 Confirm the session is really gone
try:
    fablib.get_slice(name=SLICE_NAME)
    print("⚠️ The session still appears to exist — re-run the cell above, or release it from the FABRIC Portal.")
except Exception:
    print(f"✅ Confirmed: '{SLICE_NAME}' no longer exists. You're all cleaned up.")
```

## 📖 Reference Guide: Key Concepts for Accessing National Research Computing Resources

### 🌐 FABRIC & Testbed Fundamentals (Concepts 1–10)
1. **Research Testbed** – A shared, programmable facility (FABRIC, CloudLab, …) where researchers can run real computing and networking experiments at national scale.
2. **Computing Session** – A temporary, isolated bundle of resources allocated to one experiment (FABRIC calls this a "slice").
3. **Secure Gateway Computer** – A hardened server (also called a "bastion host") that every connection must authenticate through first, before reaching any research computing node.
4. **Access Token** – A credential you download from the FABRIC Portal through a one-time browser sign-in, proving to FABRIC's systems that requests really come from you.
5. **Session State** – The life-cycle stage of a computing session: Requested → Setting Up → Ready (or Not Ready). Only "Ready" sessions are available for research work.
6. **fablib** – FABRIC's official Python toolkit for requesting, managing, and releasing computing sessions — the same toolkit this notebook just used for real.
7. **Management Address** – The network address, reachable through the secure gateway, used to connect to and control a computing node.
8. **Project Membership** – FABRIC access is granted per project; a Project Lead must approve you before you can provision real resources.
9. **Resource Request** – The processor cores, working memory (RAM), permanent storage, and starting image a researcher asks for when defining a node.
10. **Session Cleanup** – The mandatory final step of releasing resources so they return to the shared pool for other researchers.

### 💻 Computing & Workflow Fundamentals (Concepts 11–15)
11. **Temporary Cloud Workspace** – A computing session is not permanent storage; anything saved only on the session disappears once the session is released. Save your results elsewhere first (Step 2.5 did this for you).
12. **Credential Persistence** – Saving your particulars and token to Google Drive (Steps 1.2 and 1.4) means you don't have to redo browser sign-in steps every time your Colab session resets.
13. **Reproducible Experiment Description** – Writing your resource request and setup as code (rather than clicking through a series of menus) means anyone, including future you, can recreate the exact same computing session.
14. **Community Resource Ethics** – National computing systems are shared. Request only what your research actually needs, and release it promptly when finished.
15. **Official FABRIC Computing Session States** – "Ready" (StableOK) means the request finished successfully and every part of the session is working. "Not Ready" (StableError) means the request finished, but something failed and needs to be inspected.

| Plain-language state | FABRIC's technical name | Meaning |
|---|---|---|
| **Requested** | Nascent | New session, just created. |
| **Setting Up** | Configuring | Resources are being assigned and started. |
| **Ready** | StableOK | Everything requested is working. Ready for connection and scientific work. |
| **Not Ready** | StableError | Some resources started, others failed. Inspect which ones failed. |
| **Update Successful** | ModifyOK | A change to the session succeeded. |
| **Update Failed** | ModifyError | A change partially succeeded; some resources failed. |
