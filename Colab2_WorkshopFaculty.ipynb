## ==============================================================================
## MODULE 1: CHECKING HARDWARE PROPERTIES AND DATA ACQUISITION (LAYER 1: DATA INPUT)
## ==============================================================================

```python
# === MODULE 1: CHECKING HARDWARE PROPERTIES AND DATA ACQUISITION ===
import os
import sys
import json
import time
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt

# Check hardware availability
available_cpus = os.cpu_count()
print(f"[SYSTEM CHECK]: {available_cpus} CPU cores available on this machine.")

# LAYER 1: DATA INPUT
def layer_1_data_input_ecog_stream(num_electrodes=64, duration_seconds=10, sampling_rate_hz=1000):
    print("\n[LAYER 1: DATA INPUT] Initializing programmatic network channel matrix...")
    np.random.seed(42)
    total_samples = duration_seconds * sampling_rate_hz
    raw_voltages = np.random.normal(0, 25.0, (num_electrodes, total_samples))
    
    # Simulate a true baseline motor intention event
    raw_voltages[:, total_samples//2:] += np.sin(np.linspace(0, 100, total_samples//2)) * 15.0
    print(f"[LAYER 1] Data Input complete. Matrix footprint: {raw_voltages.shape} [Electrodes x Samples]")
    return raw_voltages, sampling_rate_hz
```

## ==============================================================================
## MODULE 2: COMPUTATIONAL MODEL (LAYER 2: COMPUTE CORE)
## ==============================================================================

```python
# === MODULE 2: COMPUTATIONAL MODEL (COMPUTE CORE) ===
def layer_2_compute_high_gamma_power(raw_matrix, fs, signal_gain_modifier=1.0):
    print("[LAYER 2: COMPUTE CORE] Applying zero-phase bandpass filtering [70Hz - 150Hz]...")
    
    # Structural Anomaly Check (The Mutation Playground Catchment)
    if signal_gain_modifier > 3.5:
        raise ValueError(f"CRITICAL STATE FAULT: Upstream Gain Multiplier '{signal_gain_modifier}' "
                         f"exceeds physical amplifier saturation threshold (3.5x max).")
        
    num_channels, total_samples = raw_matrix.shape
    nyquist = 0.5 * fs
    low_cutoff, high_cutoff = 70.0 / nyquist, 150.0 / nyquist
    
    b, a = butter(4, [low_cutoff, high_cutoff], btype='band')
    processed_matrix = np.zeros_like(raw_matrix)
    
    for c in range(num_channels):
        filtered = filtfilt(b, a, raw_matrix[c, :]) * signal_gain_modifier
        processed_matrix[c, :] = filtered ** 2
        
    mean_system_power = float(np.mean(processed_matrix))
    print(f"[LAYER 2] Compute complete. Integrated System Power Intensity: {mean_system_power:.4f}")
    return processed_matrix, mean_system_power
```

## ==============================================================================
## MODULE 3: VISUALIZATION DASHBOARD (LAYER 3: ANALYTICS UI)
## ==============================================================================
### FACULTY KEY: Single-Channel Dual-Axis Telemetry Render

```python
# === MODULE 3: VISUALIZATION DASHBOARD (ANALYTICS UI) ===
def layer_3_render_dashboard(raw_matrix, processed_matrix, mean_power):
    print("[LAYER 3: ANALYTICS UI] Generating dual-axis visualization array...")
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
    time_vector = np.linspace(0, raw_matrix.shape[1]/1000, raw_matrix.shape[1])
    
    # Raw signal track
    ax1.plot(time_vector, raw_matrix[0, :], color='dimgray', alpha=0.7, label='Raw Surface Potential')
    ax1.set_ylabel("Voltage ($\mu$V)")
    ax1.set_title("Neural Workflow Telemetry Assembly Dashboard")
    ax1.grid(True, linestyle='--', alpha=0.5)
    ax1.legend(loc='upper left')
    
    # Processed high-gamma envelope
    ax2.plot(time_vector, processed_matrix[0, :], color='crimson', linewidth=1.5, label='High-Gamma Power Envelope')
    ax2.axhline(y=100.0, color='darkred', linestyle=':', label='Compliance Alert Limit')
    ax2.set_ylabel("Power Intensity ($\mu$V$^2$)")
    ax2.set_xlabel("Elapsed Experiment Execution Time (Seconds)")
    ax2.grid(True, linestyle='--', alpha=0.5)
    ax2.legend(loc='upper left')
    
    plt.tight_layout()
    plt.close() # Prevents canvas clashing during automatic playground execution
    print("[LAYER 3] Visual frame rendering complete.")
    return fig
```
