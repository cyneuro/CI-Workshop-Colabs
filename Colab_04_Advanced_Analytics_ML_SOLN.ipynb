### 🧠 MODULE 0A: Feature Spaces, Unsupervised Clustering, and Classification
#### 🦾 CELL A.1: The Microelectrode Spike Sorting Cluster Engine Solution

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# 1. Generate 1000 synthetic spike waveforms representing 3 separate neurons
np.random.seed(42)
n_spikes = 1000
waveform_length = 40
t_wave = np.linspace(0, 1, waveform_length)

# Define baseline shapes for 3 different physical neurons
neuron_1 = np.sin(2 * np.pi * t_wave) * 2.0
neuron_2 = -np.sin(2 * np.pi * t_wave) * 1.5
neuron_3 = np.sin(4 * np.pi * t_wave) * 1.0

spikes = []
true_labels = []
for _ in range(n_spikes):
    cluster_id = np.random.choice([0, 1, 2])
    true_labels.append(cluster_id)
    noise = np.random.normal(0, 0.4, waveform_length)
    if cluster_id == 0:
        spikes.append(neuron_1 + noise)
    elif cluster_id == 1:
        spikes.append(neuron_2 + noise)
    else:
        spikes.append(neuron_3 + noise)

spikes = np.array(spikes)

# 2. Dimension Reduction via PCA down to 2 principal vectors
pca = PCA(n_components=2)
pca_features = pca.fit_transform(spikes)

# 3. Run K-Means Clustering on the low-dimensional feature matrix
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
predicted_labels = kmeans.fit_predict(pca_features)

# 4. Unified Dashboard Plot
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Left: PCA cluster space colored by predicted clusters
scatter = axes[0].scatter(pca_features[:, 0], pca_features[:, 1], c=predicted_labels, cmap='viridis', alpha=0.6)
axes[0].set_title("PCA Feature Space (Isolated Clusters)")
axes[0].set_xlabel("PC1 Vector Space")
axes[0].set_ylabel("PC2 Vector Space")
axes[0].grid(True, linestyle='--', alpha=0.5)

# Right: Overlaid clustered waveform profiles
for cluster in [0, 1, 2]:
    cluster_mask = (predicted_labels == cluster)
    mean_waveform = spikes[cluster_mask].mean(axis=0)
    axes[1].plot(mean_waveform, label=f"Neuron Cluster {cluster}", linewidth=2.5)

axes[1].set_title("Identified Mean Waveform Signatures")
axes[1].set_xlabel("Time Index (Bins)")
axes[1].set_ylabel("Voltage (Arbitrary Units)")
axes[1].legend()
plt.tight_layout()
plt.show()
```

#### 🧠 CELL A.2: Non-Linear Intent Decoding & Behavior Classification Solution

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay

# 1. Create a synthetic behavior intent dataset (500 trials, 10 feature values)
X = np.random.randn(500, 10)
y = np.random.choice([0, 1], size=500) # Intent types: e.g., Reach Left (0) vs Reach Right (1)

# Inject structural predictability to make the classifier function properly
X[y == 1, :3] += 1.2 

# 2. Split dataset with a rigorous 80/20 partition matrix
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

# 3. Train ML Classifier
clf = RandomForestClassifier(n_estimators=50, random_state=42)
clf.fit(X_train, y_train)

# 4. Evaluate performance
predictions = clf.predict(X_test)
print("📋 Classifier Performance Metrics Report:\n")
print(classification_report(y_test, predictions))

# Generate Confusion Matrix Plot
cm = confusion_matrix(y_test, predictions)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Reach Left', 'Reach Right'])
disp.plot(cmap='Blues')
plt.title("Decoder Behavioral Intent Confusion Matrix")
plt.show()
```
