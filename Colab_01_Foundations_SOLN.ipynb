### 🧠 MODULE 0A: Microcredential I — Sensor Physics & Basic Signal Processing
#### 🦾 CELL A.1: Synthetic EEG and fMRI Signal Generation Solution

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate Synthetic EEG Signal (10 Hz Alpha Rhythm + Noise)
fs_eeg = 500  # Sampling frequency: 500 Hz
t_eeg = np.linspace(0, 5, fs_eeg * 5, endpoint=False)  # 5 seconds
alpha_wave = np.sin(2 * np.pi * 10 * t_eeg)  # 10 Hz pure sine wave
noise_eeg = np.random.normal(0, 0.5, size=len(t_eeg))  # Gaussian noise
eeg_signal = alpha_wave + noise_eeg

# 2. Generate Synthetic fMRI BOLD Signal (Slow Hemodynamic Response)
fs_fmri = 0.5  # Sampling frequency: 0.5 Hz (TR = 2 seconds)
t_fmri = np.linspace(0, 100, int(100 * fs_fmri), endpoint=False)  # 100 seconds
bold_signal = np.sin(2 * np.pi * 0.02 * t_fmri) * np.exp(-t_fmri / 40) + np.random.normal(0, 0.1, size=len(t_fmri))

# 3. Plotting the results
fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].plot(t_eeg, eeg_signal, color='teal', label='Raw EEG (Alpha + Noise)')
axes[0].plot(t_eeg, alpha_wave, color='red', linestyle='--', alpha=0.8, label='True 10Hz Alpha')
axes[0].set_title("Synthetic EEG Signal (High Temporal Resolution)")
axes[0].set_xlabel("Time (seconds)")
axes[0].set_ylabel("Voltage (μV)")
axes[0].legend()

axes[1].plot(t_fmri, bold_signal, color='purple', marker='o', label='BOLD Signal')
axes[1].set_title("Synthetic fMRI BOLD Signal (Low Temporal Resolution)")
axes[1].set_xlabel("Time (seconds)")
axes[1].set_ylabel("Arbitrary Intensity")
axes[1].legend()

plt.tight_layout()
plt.show()
```

#### 🧠 CELL A.2: Volatile Memory and OOM Safety Evaluation Solution

```python
import sys

# Simulate voxel dimensions for a standard 3D fMRI volume
x, y, z = 64, 64, 30
timepoints = 1200

# Create synthetic float64 array
fmri_volume = np.zeros((x, y, z, timepoints), dtype=np.float64)
byte_size_64 = fmri_volume.nbytes
mb_size_64 = byte_size_64 / (1024 ** 2)

# Downcast to float32 to optimize footprint
fmri_volume_32 = fmri_volume.astype(np.float32)
byte_size_32 = fmri_volume_32.nbytes
mb_size_32 = byte_size_32 / (1024 ** 2)

print(f"📐 Array Shape: {fmri_volume.shape}")
print(f"💾 Float64 Size: {mb_size_64:.2f} MB")
print(f"⚡ Optimized Float32 Size: {mb_size_32:.2f} MB")
print(f"📉 Memory Savings: {((mb_size_64 - mb_size_32) / mb_size_64)*100:.1f}%")
```
